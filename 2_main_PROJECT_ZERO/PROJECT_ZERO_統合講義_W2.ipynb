### 〈3つの問い — 判断の理由〉

正解を覚えるより、**なぜそう決めたか** が大事です。上の差分図がそう描かれている理由は次の通りです。

**問1：新しいタブを作る**

W1の「登録」タブは **人が手で打ち込む** 画面でした。クロールは **URLを1本渡して機械に任せる** 操作で、
入力するものも待ち時間もまったく違います。**操作の性質が違うものは、タブを分けたほうが迷いません。**

さらに実利があります。タブを分ければ **W1の「登録」タブのコードを1行も消さずに済みます**。
（今週は名前を `手動登録` に変えるだけ。これが「育てる開発」です）

**問2：進捗を必ず出す**

進捗を出さないと、処理中の画面は **フリーズしたのと見分けがつきません**。
ユーザーは不安になって実行ボタンを連打し、同じクロールが二重に走ります。
Streamlit なら `st.progress()` や `st.spinner()` で「動いている」ことを見せられます。

**問3：失敗したURLまで出す**

クロールは **必ず一定数失敗します**（URLの打ち間違い、相手サイトの障害、タイムアウト）。
ここで失敗を黙って捨てると、あとで **「登録したはずのページが検索に出てこない」** という
原因の分からない不具合になります。

だから Step 3 で作るクローラーは、失敗しても例外で止まらず
**`crawl_status` に `failed` を入れた辞書を返す** 設計にします。
——**この画面設計の判断が、そのまま関数の設計につながっています。**

Python の `set`（集合）を使うと、「AにあってBにないもの」を引き算で出せます。


辞書を `set()` に入れると **キーだけの集合** になります。これを使えば、
「クロール済みデータにあって、W1のデータには無いキー」＝ **W2で増えたキー** が一発で出せます。

### `crawl_status` があると何が嬉しいのか

配布した `pages_w2.json` には、**わざと1件だけ失敗データ** を入れてあります。
クロールは、URLの打ち間違いや相手サイトの障害で **必ず一定数失敗します**。

そのとき失敗を黙って捨てると、あとで **「登録したはずのページが検索に出てこない」** という、
原因の分からない不具合になります。次のコードで、その1件を探してみましょう。


````python
# ── 復習③：for でリストを1件ずつ確認する ──
# この「for page in pages:」が、このあと作る検索ロジックの骨格になる

success = []
failed  = []

for page in pages:
    # .get() は「キーが無ければ既定値を返す」取り出し方
    # W1のデータには crawl_status が無いので "manual"（手動登録）扱いにする
    status = page.get("crawl_status", "manual")

    if status == "success":
        success.append(page)
    elif status == "failed":
        failed.append(page)

print(f"手動登録 : {len(pages) - len(success) - len(failed)} 件")
print(f"取得成功 : {len(success)} 件")
print(f"取得失敗 : {len(failed)} 件")
print()

# ── 失敗した1件を、キーと値をすべて表示して確認する ──
print("── 失敗したページの中身 ──")
for key, value in failed[0].items():
    print(f"  {key}: {value}")
````

今週クローラーが `full_text`（本文）を集めてくれるので、**調べる対象を本文まで広げられる** ようになります。

### 〈クローリングとスクレイピングの違い〉

「Webから情報を取ってくる」技術には、よく似た2つの言葉があります。

| | クローリング（crawling） | スクレイピング（scraping） |
|---|---|---|
| やること | **リンクを辿って巡回**し、ページを集めて回る | **1ページのHTMLから必要な情報を抜き出す** |
| 関心 | 「**次にどのURL**を見るか」（探索） | 「**どのタグから何**を取るか」（抽出） |
| たとえると | 図書館を歩き回って本を見つける人 | 見つけた本から必要な箇所を書き写す人 |
| Googleでは | Googlebot がWeb中を巡回する | 集めたページから情報を抜き出す |

実際のクローラーは、この **2つを組み合わせて** 動いています。

```
巡回して次のURLを見つける（クローリング）
        ↓
そのページから情報を抜き出す（スクレイピング）
        ↓
また次のリンクへ…（くり返し）
```

> 📌 **正直に言うと、今週作るのは「スクレイピング」寄りです。**

## 3-0：コードを書く前に — クロールのマナー（必ず読んでください）

**次のセルから、実際に外部のWebサイトへアクセスします。**

ここから先は自分のPCの中だけの話ではありません。クローラーは
**相手のサーバーに本当にリクエストを送るプログラム** で、**他人のサイトに迷惑をかけうる操作** です。
手を動かす前に、次の4つを頭に入れてください。

| 守ること | なぜ | どうする |
|----------|------|----------|
| **アクセス間隔を空ける** | 連続リクエストは相手サーバーへの攻撃と区別がつかない | 1件ごとに `time.sleep(1)` を入れる |
| **robots.txt を確認する** | サイト側が「ここは巡回しないで」と表明している場所がある | `https://（サイト）/robots.txt` を開いて確認 |
| **利用規約を確認する** | 規約でデータ収集を禁止しているサイトがある | 特に会員制サイト・SNSは要注意 |
| **取得先を選ぶ** | 収集した本文には著作権がある | 研修では **自分のLP・自社の公開ページ・受講者どうしのLP** を中心に |

**ループで複数ページを回すときは、必ずウェイトを入れましょう：**

```python
import time

for url in url_list:
    result = crawl_url(url)
```

Webページの中には、他のページへのリンクがあります。

1. `a` タグを探す

> 🕷 **ここが「本物のクローラー」への入口です**

````python
# リンクを保存するための空リスト
links = []

# href属性を持つaタグをすべて探す
for a in soup.find_all("a", href=True):
    
    href = a["href"]
    
    # 外部リンク（httpから始まるもの）だけ保存する
    if href.startswith("http"):
        links.append(href)

print("リンク数:", len(links))
print("リンク一覧（先頭3件）:", links[:3])
````

## 3-9：fetch_page() にまとめよう

ここまでやった「Webページ取得」の部分を、  
`fetch_page()` という関数にまとめます。

この関数の役割はシンプルです。

```text
URLを受け取る
↓
HTMLを返す
```

失敗した場合は `None` を返すようにします。


````python
from typing import Optional

def fetch_page(url: str, timeout: int = 10) -> Optional[str]:
    """
    指定URLのHTMLを取得する。

    Args:
        url: 取得対象URL
        timeout: タイムアウト秒数

    Returns:
        HTML文字列。失敗時は None
    """
    try:
        # User-Agentを指定してアクセスする
        headers = {"User-Agent": "Tech0SearchBot/1.0 (Educational Purpose)"}
        
        # Webページにアクセス
        resp = requests.get(url, headers=headers, timeout=timeout)
        
        # エラーコードなら例外を発生させる
        resp.raise_for_status()
        
        # 文字コードを推定して設定
        resp.encoding = resp.apparent_encoding
        
        # HTML全文を返す
        return resp.text

    except requests.RequestException as e:
        # 取得失敗時は None を返す
        print(f"❌ 取得エラー: {e}")
        return None
````

## 3-10：parse_html() にまとめよう

次に、HTMLを解析して情報を取り出す部分を  
`parse_html()` という関数にまとめます。

この関数の役割は

```
HTMLを受け取る
↓
ページ情報の辞書を返す
```

です。


````python
import re

def parse_html(html: str, url: str) -> dict:
    """
    HTMLを解析してページ情報を抽出する。

    Args:
        html: HTML文字列
        url: 元URL

    Returns:
        抽出した情報の辞書
    """
    # HTMLをBeautifulSoupで解析
    soup = BeautifulSoup(html, "html.parser")

    # 不要タグを除去
    for tag in soup(["script", "style", "nav", "footer", "header"]):
        tag.decompose()

    # タイトル取得
    title = "No Title"
    if soup.find("title"):
        title = soup.find("title").get_text().strip()
    elif soup.find("h1"):
        title = soup.find("h1").get_text().strip()

    # meta description
    description = ""
    meta = soup.find("meta", attrs={"name": "description"})
    if meta and meta.get("content"):
        description = meta["content"][:200]

    # meta keywords
    keywords = []
    meta_kw = soup.find("meta", attrs={"name": "keywords"})
    if meta_kw and meta_kw.get("content"):
        keywords = [kw.strip() for kw in meta_kw["content"].split(",")][:10]

    # 本文テキスト
    elems = soup.find_all(["p", "h1", "h2", "h3", "h4", "h5", "h6", "li", "td"])
    full_text = " ".join(e.get_text().strip() for e in elems)
    full_text = re.sub(r"\s+", " ", full_text).strip()

    # リンク
    links = [
        a["href"]
        for a in soup.find_all("a", href=True)
        if a["href"].startswith("http")
    ][:20]

    return {
        "url": url,
        "title": title,
        "description": description,
        "keywords": keywords,
        "full_text": full_text,
        "links": links,
        "word_count": len(full_text.split()),
        "crawled_at": datetime.now().isoformat(),
        "crawl_status": "success",
    }
````

## 3-11：crawl_url() にまとめよう

最後に、

- `fetch_page()`
- `parse_html()`

をつないで、  
`crawl_url()` を作ります。

この関数が、**クローラーの入口** になります。

```text
URLを渡す
↓
HTML取得
↓
HTML解析
↓
辞書を返す
```


````python
def crawl_url(url: str) -> dict:
    """
    URLをクロールして情報を返す（fetch → parse のワンストップ）。

    Args:
        url: クロール対象URL

    Returns:
        ページ情報の辞書（失敗時も crawl_status で判別可能）
    """
    # HTML取得
    html = fetch_page(url)

    # 取得失敗時
    if not html:
        return {
            "url": url,
            "crawl_status": "failed",
            "crawled_at": datetime.now().isoformat(),
            "error": "Failed to fetch page",
        }

    try:
        # 取得成功時は解析して返す
        return parse_html(html, url)

    except Exception as e:
        # 解析失敗時
        return {
            "url": url,
            "crawl_status": "error",
            "crawled_at": datetime.now().isoformat(),
            "error": str(e),
        }
````

## 3-12：完成したクローラーを試してみよう

最後に、完成した `crawl_url()` を実行して、

このコードでは、

2. 成功したらタイトルや文字数を表示する  
3. 失敗したらエラーを表示する

````python
result = crawl_url("https://bootcamp.tech0-jp.com/")

if result.get("crawl_status") == "success":
    print("✅ クロール成功!")
    print(f"📄 タイトル: {result['title']}")
    print(f"📝 説明: {result['description'][:100]}...")
    print(f"📊 文字数: {result['word_count']}語")
    print(f"🔗 リンク数: {len(result['links'])}件")
else:
    print(f"❌ クロール失敗: {result.get('error')}")
````

### 🕷 発展課題：「本物のクローラー」にしてみる（難易度高め）

Step 3 で作った `crawl_url()` は、**渡されたURLを1件処理するだけ** でした。
3-7 で取り出した `links` を **次のクロール対象に回す** と、リンクを辿って自動で巡回する
＝ **本当の意味でのクローラー** になります。

```python
import time   # ← アクセス間隔を空けるために必要

def crawl_recursive(start_url: str, max_pages: int = 5) -> list:
    visited = set()          # すでに訪れたURL（同じページを2度取らない）
    queue   = [start_url]    # これから訪れるURL
    results = []

    while queue and len(results) < max_pages:   # ← max_pages が歯止め
        url = queue.pop(0)
        if url in visited:
            continue
        visited.add(url)

        result = crawl_url(url)
        results.append(result)

        # 取り出した links を次のクロール対象に足す ← ここが「巡回」
        for link in result.get("links", []):
            if link not in visited:
                queue.append(link)

        time.sleep(1)        # ← マナー。必ず入れる

    return results
```

> ⚠️ **必ず歯止めを付けてください。** リンクは際限なく広がるので、
> `max_pages`（何件まで）を付けないと **止まらないプログラム** になります。
> 「同じドメインだけ辿る」という条件を足すとさらに安全です。
>
> ```python
> from urllib.parse import urlparse
> if urlparse(link).netloc == urlparse(start_url).netloc:   # 同じドメインだけ
>     queue.append(link)
> ```
>
> 実行する前に、最終セルの **「クロールのマナー」** をもう一度読んでください。
> 巡回型は自動で件数が増えるぶん、相手サーバーへの負荷も一気に増えます。
